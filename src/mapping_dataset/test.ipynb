```python
import json
import pandas as pd

path = r"C:\Users\tahay\OneDrive\Desktop\Attack-Progression-Prediction\zeek_mappingDB.json"

with open(path, "r", encoding="utf-8") as f:
    zeek_db = json.load(f)

zeek_df = pd.json_normalize(zeek_db["mappings"])

zeek_df.head()
```

```python
fields_by_log = {}

for _, row in zeek_df.iterrows():
    log = row["log_type"]
    print(log)
```

```python

if log not in fields_by_log:
    fields_by_log[log] = set()

    for field in row["required_fields"]:
        if isinstance(field, list):
            fields_by_log[log].update(field)
        else:
            fields_by_log[log].add(field)


for log, fields in fields_by_log.items():
    fields_by_log[log] = sorted(fields)


print(fields_by_log)
```

```python
import pandas as pd

# 1. Handle log_type lists: explode so each log type gets its own row
#    (e.g., ['http.log', 'files.log'] becomes two separate rows)
df_exploded = zeek_df.explode("log_type")

# 2. Now explode the required_fields lists
df_fields = df_exploded.explode("required_fields")

# 3. Group by log_type and collect unique required fields
unique_fields = (
    df_fields.groupby("log_type")["required_fields"]
    .apply(lambda x: sorted(x.dropna().unique()))
)

# 4. Pretty print
for log_type, fields in unique_fields.items():
    print(f"\n=== {log_type} ({len(fields)} unique fields) ===")
    for f in fields:
        print(f"  - {f}")

# 5. (Optional) Convert to a clean DataFrame for export
unique_fields_df = unique_fields.reset_index(name="unique_required_fields")
print("\n", unique_fields_df)
```
