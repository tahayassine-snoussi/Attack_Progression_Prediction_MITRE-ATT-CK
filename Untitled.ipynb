```python
import csv


INPUT_DB = "mapping_dataset.csv"
OUTPUT_DB = "wazuh_mappingDB_lab.csv"


techniques = {
    'T1078',
    'T1110',
    'T1110.003',

    'T1059',
    'T1059.001',
    'T1059.004',

    'T1547.001',
    'T1053.005',
    'T1543.003',
    'T1136.001',
    'T1098',
    'T1546.004',

    'T1548',
    'T1548.003',
    'T1068',

    'T1552.001',
    'T1003',
    'T1003.001',
    'T1003.008',

    'T1082',
    'T1033',
    'T1083',
    'T1087',
    'T1087.002',
    'T1069.002',
    'T1018',
    'T1016',
    'T1049',

    'T1021',
    'T1021.001',
    'T1021.002',
    'T1021.004',

    'T1027',
    'T1055',
    'T1562.001',
    'T1070.004',

    'T1105',
    'T1071.001',
    'T1071.004',

    'T1005',
    'T1119',
    'T1560',
    'T1560.001',

    'T1041',
    'T1048',
}


with open(
    INPUT_DB,
    "r",
    encoding="utf-8",
    newline=""
) as infile:

    reader = csv.DictReader(infile)

    rows = list(reader)
    fieldnames = reader.fieldnames


filtered_rows = []

for row in rows:

    technique_id = row.get("technique_id")

    if technique_id in techniques:
        filtered_rows.append(row)


with open(
    OUTPUT_DB,
    "w",
    encoding="utf-8",
    newline=""
) as outfile:

    writer = csv.DictWriter(
        outfile,
        fieldnames=fieldnames
    )

    writer.writeheader()

    writer.writerows(filtered_rows)


print("=" * 70)
print("WAZUH LAB DATABASE CREATED")
print("=" * 70)

print(f"Original mappings : {len(rows)}")
print(f"Filtered mappings : {len(filtered_rows)}")
print(f"Output            : {OUTPUT_DB}")

print("\nTechniques included:")

found_techniques = set(
    row.get("technique_id")
    for row in filtered_rows
)

for technique in sorted(techniques):

    count = sum(
        1
        for row in filtered_rows
        if row.get("technique_id") == technique
    )

    print(
        f"{technique:12} -> {count}"
    )


print("\nMissing techniques:")

missing = techniques - found_techniques

if missing:
    for technique in sorted(missing):
        print(f"  {technique}")
else:
    print("  None")
```
