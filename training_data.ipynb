```python
import json
from datetime import datetime


# ============================================================
# 1. Helper functions
# ============================================================

def get_attack_id(external_references):
    """
    Extract MITRE ATT&CK external_id from STIX external_references.
    Example:
        T1059
        T1059.001
        T1105
    """
    if not isinstance(external_references, list):
        return None

    for ref in external_references:
        if ref.get("source_name") == "mitre-attack":
            return ref.get("external_id")

    return None


def get_mitre_group_id(external_references):
    """
    Extract the MITRE ATT&CK group ID, e.g. G0001.
    """
    if not isinstance(external_references, list):
        return None

    for ref in external_references:
        if ref.get("source_name") == "mitre-attack":
            external_id = ref.get("external_id")

            if external_id and external_id.startswith("G"):
                return external_id

    return None


# ============================================================
# 2. Make sure techniques have their ATT&CK IDs
# ============================================================

if "technique_id" not in techniques_df.columns:
    techniques_df["technique_id"] = (
        techniques_df["external_references"]
        .apply(get_attack_id)
    )


# ============================================================
# 3. Add MITRE group IDs
# ============================================================

groups_df = groups_df.copy()

groups_df["group_id"] = (
    groups_df["external_references"]
    .apply(get_mitre_group_id)
)


# ============================================================
# 4. Build technique lookup
# ============================================================

technique_lookup = {}

for _, row in techniques_df.iterrows():

    technique_id = row["technique_id"]

    if not technique_id:
        continue

    # Prefer your existing Wazuh/Zeek mapping for tactic
    tactic = None

    if technique_id in supported_techniques:
        tactic = supported_techniques[technique_id].get("tactic")

    # Make sure tactics are always a list
    if tactic:
        if isinstance(tactic, list):
            tactics = tactic
        else:
            tactics = [tactic]
    else:
        tactics = []

    technique_lookup[technique_id] = {
        "technique_name": row["name"],
        "stix_id": row["id"],
        "tactics": tactics
    }


# ============================================================
# 5. Build group lookup
# ============================================================

group_lookup = {}

for _, row in groups_df.iterrows():

    group_id = row["group_id"]

    if not group_id:
        continue

    group_lookup[group_id] = {
        "group_name": row["name"],
        "stix_id": row["id"]
    }

# ============================================================
# 6. Build lookup dictionaries
# ============================================================

group_by_stix_id = {
    row["id"]: row
    for _, row in groups_df.iterrows()
}

technique_by_stix_id = {
    row["id"]: row
    for _, row in techniques_df.iterrows()
}


# ============================================================
# Build normalized Group → Technique relationships
# ============================================================

group_technique_relationships = []

for _, row in relationships_df.iterrows():

    source_ref = row["source_ref"]
    target_ref = row["target_ref"]

    group = group_by_stix_id.get(source_ref)
    technique = technique_by_stix_id.get(target_ref)

    if group is None or technique is None:
        continue

    group_id = group["group_id"]
    technique_id = technique["technique_id"]

    if not group_id or not technique_id:
        continue

    relationship = {
        "relationship_id": row["id"],
        "relationship_type": row["relationship_type"],

        "group_id": group_id,
        "group_stix_id": source_ref,

        "technique_id": technique_id,
        "technique_stix_id": target_ref,

        "source": {
            "type": "MITRE_ATT&CK_STIX",
            "confidence": 1.0
        }
    }

    if (
        "description" in row.index
        and isinstance(row["description"], str)
        and row["description"].strip()
    ):
        relationship["description"] = row["description"]

    group_technique_relationships.append(relationship)
# ============================================================
# 7. Build technique-centric relationships
#
# Source of truth:
#     group_technique_relationships
#
# For each supported input technique:
#
#     input technique
#          ↓
#     groups using it
#          ↓
#     other techniques used by those groups
#          ↓
#     count shared groups
# ============================================================

technique_relations = {}


# ------------------------------------------------------------
# Build:
#
# technique → groups
#
# directly from normalized relationships
# ------------------------------------------------------------

technique_to_groups = {}

for relationship in group_technique_relationships:

    if relationship["relationship_type"] != "uses":
        continue

    technique_id = relationship["technique_id"]
    group_id = relationship["group_id"]

    if technique_id not in technique_to_groups:
        technique_to_groups[technique_id] = set()

    technique_to_groups[technique_id].add(group_id)


# ------------------------------------------------------------
# Build:
#
# group → techniques
#
# directly from normalized relationships
# ------------------------------------------------------------

group_to_techniques = {}

for relationship in group_technique_relationships:

    if relationship["relationship_type"] != "uses":
        continue

    group_id = relationship["group_id"]
    technique_id = relationship["technique_id"]

    if group_id not in group_to_techniques:
        group_to_techniques[group_id] = set()

    group_to_techniques[group_id].add(technique_id)


# ------------------------------------------------------------
# Generate technique-centric knowledge
# ------------------------------------------------------------

for input_technique in technique_lookup.keys():

    # --------------------------------------------------------
    # Groups using the input technique
    # --------------------------------------------------------

    groups_using_input = sorted(
        technique_to_groups.get(
            input_technique,
            set()
        )
    )


    # --------------------------------------------------------
    # Find techniques used by those groups
    # --------------------------------------------------------

    related = {}

    for group_id in groups_using_input:

        techniques_used_by_group = group_to_techniques.get(
            group_id,
            set()
        )

        for other_technique in techniques_used_by_group:

            # Don't relate technique to itself
            if other_technique == input_technique:
                continue

            if other_technique not in related:

                related[other_technique] = {
                    "group_count": 0,
                    "groups": []
                }

            if group_id not in related[other_technique]["groups"]:

                related[other_technique]["groups"].append(
                    group_id
                )


    # --------------------------------------------------------
    # Calculate group counts
    # --------------------------------------------------------

    for other_technique in related:

        related[other_technique]["groups"].sort()

        related[other_technique]["group_count"] = len(
            related[other_technique]["groups"]
        )


    # --------------------------------------------------------
    # Sort strongest relationships first
    # --------------------------------------------------------

    related = dict(
        sorted(
            related.items(),
            key=lambda item: (
                -item[1]["group_count"],
                item[0]
            )
        )
    )


    # --------------------------------------------------------
    # Store
    # --------------------------------------------------------

    technique_relations[input_technique] = {

        "group_count": len(groups_using_input),

        "groups": groups_using_input,

        "related_techniques": related
    }
# ============================================================
# 8. Build the final database
# ============================================================

database = {

    "metadata": {
        "dataset_name": "attack_progression_knowledge",
        "version": "0.1.0",
        "created_at": datetime.utcnow().isoformat() + "Z",

        "primary_source": {
            "name": "MITRE ATT&CK",
            "format": "STIX 2.1",
            "domain": "enterprise-attack"
        },

        "technique_vocabulary": {
            "source": "Wazuh/Zeek semantic mapping",
            "supported_technique_count": len(technique_lookup)
        },

        "relationship_semantics": {
            "uses": "Group is associated with / uses a technique",
            "important_note": (
                "MITRE ATT&CK uses relationships do not imply "
                "chronological technique ordering."
            )
        }
    },


    # ========================================================
    # TECHNIQUES
    # ========================================================

    "techniques": technique_lookup,


    # ========================================================
    # GROUPS
    # ========================================================

    "groups": group_lookup,


    # ========================================================
    # RAW / NORMALIZED GROUP → TECHNIQUE EDGES
    # ========================================================

    "group_technique_relationships": group_technique_relationships,


    # ========================================================
    # TECHNIQUE → RELATED TECHNIQUES
    # ========================================================

    "technique_relations": technique_relations,


    # ========================================================
    # SEQUENCE EVIDENCE
    #
    # Empty for now because STIX does not provide the
    # chronological ordering we need.
    # ========================================================

    "sequence_evidence": []
}


# ============================================================
# 9. Save JSON
# ============================================================

output_file = "attack_progression_knowledge.json"

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(
        database,
        f,
        indent=2,
        ensure_ascii=False
    )


print(f"Database saved to: {output_file}")

print()
print("Statistics")
print("--------------------------------")
print("Techniques:", len(database["techniques"]))
print("Groups:", len(database["groups"]))
print(
    "Group-Technique relationships:",
    len(database["group_technique_relationships"])
)
print(
    "Technique relation entries:",
    len(database["technique_relations"])
)
```

```python
import json

# ============================================================
# Load database
# ============================================================

with open(
    "attack_progression_knowledge.json",
    "r",
    encoding="utf-8"
) as f:
    db = json.load(f)


# ============================================================
# Input technique
# ============================================================

input_technique = "T1048"


# ============================================================
# Get technique information
# ============================================================

technique = db["techniques"].get(input_technique)

if technique is None:
    print(f"Technique {input_technique} not found.")
    exit()


print("=" * 70)
print("INPUT TECHNIQUE")
print("=" * 70)

print(f"ID      : {input_technique}")
print(f"Name    : {technique['technique_name']}")
print(f"Tactics : {', '.join(technique['tactics'])}")

# ============================================================
# Get technique relation record
# ============================================================

technique_relation = db["technique_relations"].get(
    input_technique
)

if technique_relation is None:
    print(
        f"No relation data found for {input_technique}"
    )
    exit()


# ============================================================
# Groups using input technique
# ============================================================

groups_using_technique = technique_relation["groups"]

print()
print("=" * 70)
print("GROUPS USING THE INPUT TECHNIQUE")
print("=" * 70)

for group_id in groups_using_technique:

    group = db["groups"].get(group_id)

    if group:
        print(
            f"{group_id} | "
            f"{group['group_name']}"
        )

print()
print(
    f"Total groups: "
    f"{len(groups_using_technique)}"
)


# ============================================================
# Related techniques
# ============================================================

related_techniques = technique_relation[
    "related_techniques"
]

print()
print("=" * 70)
print("RELATED TECHNIQUES")
print("=" * 70)

for technique_id, relation in related_techniques.items():

    technique_info = db["techniques"].get(
        technique_id
    )

    if not technique_info:
        continue

    print(
        f"{technique_id:<12} | "
        f"{technique_info['technique_name']:<55} | "
        f"Groups: {relation['group_count']}"
    )

print()
print(
    f"Total related techniques: "
    f"{len(related_techniques)}"
)
```

```python
import json

with open(
    "attack_progression_knowledge.json",
    "r",
    encoding="utf-8"
) as f:
    db = json.load(f)


target = "T1204.002"


# ============================================================
# 1. Raw relationships
# ============================================================

raw_groups = set()

for rel in db["group_technique_relationships"]:

    if (
        rel["technique_id"] == target
        and rel["relationship_type"] == "uses"
    ):
        raw_groups.add(rel["group_id"])


print("=" * 70)
print("RAW RELATIONSHIPS")
print("=" * 70)

print("Groups:", len(raw_groups))
print(sorted(raw_groups))


# ============================================================
# 2. technique_relations
# ============================================================

relation = db["technique_relations"].get(target, {})

print()
print("=" * 70)
print("TECHNIQUE RELATIONS")
print("=" * 70)

print("Related techniques:", len(relation))

# Count groups associated with the input technique
# from the related-technique section itself
relation_groups = set()

for technique_id, info in relation.items():

    for group_id in info.get("groups", []):
        relation_groups.add(group_id)


print("Groups represented:", len(relation_groups))
print(sorted(relation_groups))


# ============================================================
# 3. Compare
# ============================================================

print()
print("=" * 70)
print("COMPARISON")
print("=" * 70)

print("Raw groups:", len(raw_groups))
print("Relation groups:", len(relation_groups))

print("Missing from technique_relations:")
print(sorted(raw_groups - relation_groups))

print()
print("Extra in technique_relations:")
print(sorted(relation_groups - raw_groups))
```
